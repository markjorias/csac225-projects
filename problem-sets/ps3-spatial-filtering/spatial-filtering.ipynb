**Adding Gaussian Noise**

```python
# Adding gaussian noise

# Load image in grayscale
image_path = "ningning-1080.jpg"
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Add Gaussian noise to the image
noise = np.random.normal(0, 1, image.shape).astype(np.uint8)
noisy_image = cv2.add(image, noise)

# Display the noisy image
plt.imshow(cv2.cvtColor(noisy_image, cv2.COLOR_BGR2RGB))
plt.title("Noisy Image")
plt.axis("off")
plt.show()
```

### **Problem 1:** Noise Reduction Using Spatial Filters

##### **Mean filtering**

Mean filtering is a widely employed technique within the realm of image processing, serving as a crucial tool for noise reduction and smoothing. By adopting spatial filtering principles, mean filtering aids in eliminating random variations or noise present in an image, all while retaining its essential features.

```python
def mean_filter(image, kernel_size):
    # Pad the image to handle borders using zero-padding
    # This ensures that the filter can be applied to border pixels as well
    pad = kernel_size // 2
    padded_image = cv2.copyMakeBorder(image, pad, pad, pad, pad, cv2.BORDER_CONSTANT)
    
    # Create an empty image to store the filtered result
    filtered_image = np.zeros_like(image)
    
    # Iterate over each pixel in the image
    for y in range(image.shape[0]):
        for x in range(image.shape[1]):
            # Extract the local window centered at the current pixel
            window = padded_image[y:y+kernel_size, x:x+kernel_size]
            # Calculate the mean value of the window
            mean_value = np.mean(window)
            # Assign the mean value to the corresponding pixel in the filtered image
            filtered_image[y, x] = mean_value
    
    return filtered_image

# Load the image in grayscale mode
#image_path = "ningning-1080.jpg"
image = noisy_image

# Define the kernel size for the mean filter
kernel_size = 4

# Apply the mean filter to the image
mean_filtered_image = mean_filter(image, kernel_size)

# Display the original and filtered images
plt.figure(figsize=(15, 10))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(mean_filtered_image, cmap='gray')
plt.title("Mean Filtered Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

##### **Median filtering**

Median filtering is a type of nonlinear filter used in computer vision that replaces the value of each pixel in an image with the median value of the neighboring pixels. It is a simple and effective technique for reducing noise in images and can also be used for image smoothing and preserving edges.

```python
def median_filter(image, kernel_size=3):  
    # Pad the image to handle borders using zero-padding
    # This ensures that the filter can be applied to border pixels as well
    pad = kernel_size // 2
    padded_image = cv2.copyMakeBorder(image, pad, pad, pad, pad, cv2.BORDER_CONSTANT)
    
    # Create an empty image to store the filtered result
    filtered_image = np.zeros_like(image)
    
    # Iterate over each pixel in the image
    for y in range(image.shape[0]):
        for x in range(image.shape[1]):
            # Extract the local window centered at the current pixel
            window = padded_image[y:y+kernel_size, x:x+kernel_size]
            # Calculate the median value of the window
            median_value = np.median(window)
            # Assign the median value to the corresponding pixel in the filtered image
            filtered_image[y, x] = median_value
    
    return filtered_image

# Load noisy image
image = noisy_image

# Define the kernel size for the median filter
kernel_size = 5

# Apply the median filter to the image
median_filtered_image = median_filter(image, kernel_size)

# Display the original and filtered images
plt.figure(figsize=(15, 10))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(median_filtered_image, cmap='gray')
plt.title("Median Filtered Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

##### **Side by Side Comparison of Mean and Median Filter**

```python
plt.figure(figsize=(15, 10))  # Set the figure size to 15x10 inches
plt.subplot(1, 2, 1)
plt.imshow(mean_filtered_image, cmap='gray')
plt.title("Mean Filtered Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(median_filtered_image, cmap='gray')
plt.title("Median Filtered Image")
plt.axis("off")

plt.tight_layout()  # Ensure the titles fit within the figure
plt.show()
```

### **Problem 2:** Implementing Spatial Filters in OpenCV

##### **Gaussian blur**

Applying a Gaussian Blur gives central pixels more weight than surrounding regions, effectively reducing noise while preserving image structure. The essential parameter controls the filter’s scope and subsequent smoothing. Gaussian filters are excellent at removing random, subtle image noise patterns.

```python
# Load noisy image
image = noisy_image

# Apply Gaussian blur (low-pass filter) with 5x5 kernel
kernel_size = (5, 5)
gaussian = cv2.GaussianBlur(image, kernel_size, 4)

# Display Images
plt.figure(figsize=(15, 10))  # Set the figure size to 15x10 inches

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gaussian, cmap='gray')
plt.title("Gaussian Filtered Image")
plt.axis("off")

plt.tight_layout()  # Ensure the titles fit within the figure
plt.show()
```

##### **Laplacian filter**

The Laplacian is a filter that emphasizes regions with rapid intensity transitions, typically marking the boundaries of objects within an image. It’s commonly used for edge detection and can be combined with other filters (e.g., Gaussian) for better results in noisy images.

```python
# Load noisy image
image = noisy_image

# Apply Laplacian filter for edge detection
# Using a 5x5 kernel with depth of CV_16S to avoid overflow/underflow
laplacian = cv2.Laplacian(image, cv2.CV_16S, ksize=5)

# Converting result to an absolute value
laplacian2 = cv2.convertScaleAbs(laplacian)

#Display Images
plt.figure(figsize=(15, 10))  # Set the figure size to 15x10 inches

plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Filtered Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(laplacian2, cmap='gray')
plt.title("|Laplacian Filtered Image|")
plt.axis("off")

plt.tight_layout()  # Ensure the titles fit within the figure
plt.show()
```

Combined Methods for better results using Gaussian and Laplacian filters. As we can see below there are improvements in edge detection by combining both filters.

```python
laplacian_gaussian = cv2.Laplacian(gaussian, cv2.CV_16S, ksize=5)

#Display Images
plt.figure(figsize=(20, 15))  # Set the figure size to 20x15 inches

plt.subplot(1, 3, 1)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(laplacian_gaussian, cmap='gray')
plt.title("Gaussian + Laplacian")
plt.axis("off")

plt.tight_layout()  # Ensure the titles fit within the figure
plt.show()
```

##### **Sobel filters**

Another edge detection tool is the Sobel filter. Specifically, it computes the horizontal and vertical derivatives of the image intensity function, then combines them to obtain the gradient magnitude, which represents the edges in the image.

```python
# Load noisy image
image = noisy_image

# Apply Sobel filters in both horizontal and vertical directions
sobel_x = cv2.Sobel(image, cv2.CV_16S, 1, 0, ksize=5)
sobel_y = cv2.Sobel(image, cv2.CV_16S, 0, 1, ksize=5)

# Compute gradient magnitude
gradient_magnitude = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

# Converting result to an absolute value
sobel_x_abs = cv2.convertScaleAbs(sobel_x)
sobel_y_abs = cv2.convertScaleAbs(sobel_y)
gradient_magnitude_abs = cv2.convertScaleAbs(gradient_magnitude)

#Display Images
plt.figure(figsize=(15, 10))  # Set the figure size to 15x10 inches

plt.subplot(2, 3, 1)
plt.imshow(sobel_x, cmap='gray')
plt.title('Sobel X (Horizontal)')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(sobel_y, cmap='gray')
plt.title('Sobel Y (Vertical)')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(gradient_magnitude, cmap='gray')
plt.title('Gradient Magnitude')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(sobel_x_abs, cmap='gray')
plt.title('|Sobel X (Horizontal)|')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(sobel_y_abs, cmap='gray')
plt.title('|Sobel Y (Vertical)|')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(gradient_magnitude_abs, cmap='gray')
plt.title('|Gradient Magnitude|')
plt.axis('off')

plt.tight_layout()  # Ensure the titles fit within the figure
plt.show()
```

##### **Comparing Laplacian and Sobel Edge Detections**

- The Laplacian method highlights fine details but is more susceptible to noise.

- The Sobel method provides clearer edges with a smoother result due to gradient computation.

- If noise reduction is needed, Sobel filtering might be preferable, whereas Laplacian filtering can be useful for capturing finer details but may require preprocessing like noise reduction.
