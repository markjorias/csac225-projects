#### **Problem 1: Negative Image Transformation**

```python

# Load image in grayscale (0)
im_file = 'animal-img.jpg'
img = cv2.imread(im_file, 0)

# get image properties.
h, w = np.shape(img)

# iterate over the entire image and apply the negative transformation to each pixel.
for py in range(0, h):
    for px in range(0, w):
        img[py][px] = 1 - img[py][px]

cv2.imwrite('negative-output.jpg', img)

# Read original and processed images
original = cv2.imread(im_file, 0)
processed = cv2.imread('negative-output.jpg', 0)

# Stack images horizontally
result = np.hstack((original, processed))

# Display result
cv2.imshow('Original vs Processed', result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### **Problem 2: Image Thresholding**

```python
im_file = "book-img.jpg"

# Load image in grayscale (0)
img = cv2.imread(im_file, 0)

# For Thresholding
T = 120  # configurable threshold variable

# For Adaptive Thresholding (Mean & Gaussian)
blockSize = 21  # configurable block size for adaptive thresholding, should be odd
C = 80  # configurable constant for adaptive thresholding

_, thresh1 = cv2.threshold(img, T, 255, cv2.THRESH_BINARY)
_, thresh1_inv = cv2.threshold(img, T, 255, cv2.THRESH_BINARY_INV)

adt_thresh_mean = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, blockSize, C)
adt_thresh_gauss = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, blockSize, C)

# Display images in separate windows with labels
cv2.imshow('Adaptive Gaussian Threshold', adt_thresh_gauss)
cv2.imshow('Adaptive Mean Threshold', adt_thresh_mean)
cv2.imshow('Inverse Binary Threshold', thresh1_inv)
cv2.imshow('Binary Threshold', thresh1)
cv2.imshow('Original', img)

# Wait for a key press and close the windows
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### **Problem 3: Logarithmic Transformation**

```python
im_file = "cat-img.jpg"
# Load image in grayscale (0)
img = cv2.imread(im_file,0)

#Apply log transformation method
c = 255/(np.log(1 + 255))
log_image = c*np.log(img + 1)

# float value will be converted to int
log_image = np.array(log_image, dtype=np.uint8)

# Normalize the transformed image to 0-255 range
normalized = cv2.normalize(log_image, None, 0, 255, cv2.NORM_MINMAX)
normalized = normalized.astype(np.uint8)

# Display the original and processed images
cv2.namedWindow('Original Image', cv2.WINDOW_NORMAL)
cv2.namedWindow('Logarithmic Transformation', cv2.WINDOW_NORMAL)
cv2.resizeWindow('Original Image', 800, 600)
cv2.resizeWindow('Logarithmic Transformation', 800, 600)
cv2.imshow('Original Image', img)
cv2.imshow('Logarithmic Transformation', normalized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### **Problem 4: Power-Law (Gamma) Transformation**

```python
def display_gamma_correction(im_file, gamma_values):
    # Load image in grayscale
    img = cv2.imread(im_file,0)

    for gamma in gamma_values:
        adjusted_image = np.power(img / 255.0, gamma) * 255.0
        adjusted_image = adjusted_image.astype(np.uint8)

        # Display the original and processed images side by side
        result = np.hstack((img, adjusted_image))
        cv2.namedWindow(f"Original Image vs Gamma Correction (gamma = {gamma})", cv2.WINDOW_NORMAL)
        cv2.resizeWindow(f"Original Image vs Gamma Correction (gamma = {gamma})", 1500, 600)
        cv2.imshow(f"Original Image vs Gamma Correction (gamma = {gamma})", result)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

# Call the function

image = 'cat-img.jpg'
# Gamma experimental values
display_gamma_correction(image, [0.4, 1.0, 2.5])
```

#### **Problem 5: Gray Level Slicing**

```python
image = "animal-img.jpg"

# Load image in grayscale
img = cv2.imread(image, 0)

Imin = 50  # Minimum intensity threshold
Imax = 100  # Maximum intensity threshold

# Create a mask to select pixels within the specified intensity range
intensity = (img >= Imin) & (img <= Imax)

# Segment the image by setting pixels within the range to 255 (white) and others to 0 (black)
sliced_without_bg = np.where(intensity, 255, 0).astype(np.uint8)

# Segment the image by setting pixels within the range to 255 (white) and preserving the original intensity of other pixels
sliced_with_bg = np.where(intensity, 255, img).astype(np.uint8)

# Combine original image, segmented image without background, and segmented image with background for comparison
result = np.hstack((img, sliced_without_bg, sliced_with_bg))

# Display original image vs segmented images
cv2.imshow(f'Original : Without Background : With Background', result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### **Problem 6: Bit-Plane Slicing**

```python
# Load the image in grayscale
im_file = 'animal-img.jpg'
img = cv2.imread(im_file, 0)

# List to hold bit-planes
bit_planes = []

# Extract bit-planes
for bit in range(8):
    # Right-shift by bit and bitwise AND with 1 to remove all bits other than the desired bit
    bit_plane = (img >> bit) & 1
    # Multiply by 255 to display the bit-plane
    bit_planes.append(bit_plane * 255)

# Create a window for all bit planes
height, width = img.shape
bit_plane_window = np.zeros((height * 2, width * 4), dtype=np.uint8)

# Place original image and bit planes in the window
# Original image at first position
bit_plane_window[0:height, 0:width] = img

# Place bit planes
for bit in range(8):
    row = bit // 4
    col = bit % 4
    if bit == 0:  # Skip first position as it contains the original image
        continue
    y_offset = row * height
    x_offset = col * width
    bit_plane_window[y_offset:y_offset+height, x_offset:x_offset+width] = bit_planes[bit]

# Reconstruct the image using the most significant bits (5-8)
reconstructed_img = np.zeros_like(img, dtype=np.uint8)
for i in range(4, 8):  # Using bits 4-7 (equivalent to 5-8 in 1-indexed counting)
    reconstructed_img += (bit_planes[i] // 255) * (2 ** i)

# Display the windows
cv2.imshow('Original Image and Bit Planes', bit_plane_window)
cv2.imshow('Reconstructed Image', reconstructed_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
